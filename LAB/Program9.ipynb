```python
# Compare Linear and Polynomial Regression

# Step 1: Import libraries
import pandas as pd
from google.colab import files
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Step 2: Upload Excel dataset
uploaded = files.upload()

# Step 3: Read dataset
data = pd.read_excel("linear_polynomial_regression_dataset.xlsx")

print("Dataset:")
print(data)

# Step 4: Separate input and output
X = data[["Study_Hours"]]
y = data["Marks"]

# Step 5: Linear Regression
linear = LinearRegression()
linear.fit(X, y)

linear_pred = linear.predict(X)

# Step 6: Polynomial Regression
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

polynomial = LinearRegression()
polynomial.fit(X_poly, y)

poly_pred = polynomial.predict(X_poly)

# Step 7: Compare results
print("\nLinear Regression Score:",
      linear.score(X, y))

print("Polynomial Regression Score:",
      polynomial.score(X_poly, y))

# Step 8: Predict marks for 6 study hours
test = pd.DataFrame({"Study_Hours": [6]})

print("\nLinear Prediction:",
      linear.predict(test)[0])

print("Polynomial Prediction:",
      polynomial.predict(poly.transform(test))[0])
```
